```python
from sklearn.preprocessing import MinMaxScaler

# Inicializar o MinMaxScaler
scaler = MinMaxScaler()

# Selecionar colunas para normalizar (todas, exceto a última)
columns_to_normalize = df_train.columns[:-1]

# Aplicar a normalização Min-Max
df_train[columns_to_normalize] = scaler.fit_transform(df_train[columns_to_normalize])

display(df_train.head())

# Exibir as primeiras linhas do DataFrame de treino após a normalização
```

**Preparação dos dados:**

Vamos separar as colunas de features (todas, exceto a última) e a coluna target (a última coluna, `median_house_value`) do DataFrame de treino.

```python
# Separar features (X) e target (y)
X_train = df_train.iloc[:, :-1]
y_train = df_train.iloc[:, -1]

display("Features (X_train):")
display(X_train.head())
display("Target (y_train):")
display(y_train.head())

# Separar as features (X) e o target (y) nos dados de treino
```

**Treinando o Modelo com `SGDRegressor` do Scikit-learn:**

Vamos usar a classe `SGDRegressor` do scikit-learn, que implementa um regressor linear usando Stochastic Gradient Descent (SGD). O SGD é uma variação do Gradiente Descendente.

```python
from sklearn.linear_model import SGDRegressor

# Inicializar o modelo SGDRegressor
# max_iter: número máximo de passes sobre os dados de treino (épocas)
# eta0: taxa de aprendizado inicial
# random_state: para reprodutibilidade
# penalty: Regularização (opcional, 'l2' é a padrão)
# early_stopping: Para parar o treino quando a pontuação de validação não melhorar
model = SGDRegressor(max_iter=1000, eta0=0.01, random_state=42, early_stopping=False)

# Treinar o modelo
model.fit(X_train, y_train)

display("Modelo treinado com sucesso!")
display(f"Coeficientes (pesos): {model.coef_}")
display(f"Interceptação: {model.intercept_}")

# Treinar o modelo SGDRegressor
```

**Avaliando o Modelo nos Dados de Treino (MAE e MSE):**

Vamos calcular duas métricas comuns para avaliar modelos de regressão:

- **Erro Médio Absoluto (MAE):** É a média dos valores absolutos dos erros. Mede a magnitude média dos erros sem considerar sua direção.
- **Erro Quadrático Médio (MSE):** É a média dos quadrados dos erros. Penaliza erros maiores mais severamente do que erros menores. Sua raiz quadrada (RMSE) é frequentemente usada, pois está na mesma unidade da variável target.

```python
from sklearn.metrics import mean_absolute_error, mean_squared_error

# Fazer previsões nos dados de treino
y_train_pred = model.predict(X_train)

# Calcular MAE
mae_train = mean_absolute_error(y_train, y_train_pred)

# Calcular MSE
mse_train = mean_squared_error(y_train, y_train_pred)

display(f"MAE nos dados de treino: {mae_train}")
display(f"MSE nos dados de treino: {mse_train}")

# Calcular MAE e MSE nos dados de treino
```

```python
# Preparar os dados de validação (mesmas transformações que os dados de treino)

# Remover a coluna 'ocean_proximity'
if "ocean_proximity" in df_val:
  del df_val["ocean_proximity"]

# Separar features (X_val) e target (y_val)
X_val = df_val.iloc[:, :-1]
y_val = df_val.iloc[:, -1]

# Aplicar o mesmo MinMaxScaler ajustado nos dados de treino às features de validação
# Apenas transformamos, não ajustamos novamente
X_val[X_val.columns] = scaler.transform(X_val[X_val.columns])

display("Features de validação (X_val):")
display(X_val.head())
display("Target de validação (y_val):")
display(y_val.head())

# Fazer previsões nos dados de validação
y_val_pred = model.predict(X_val)

# Calcular MAE e MSE nos dados de validação
mae_val = mean_absolute_error(y_val, y_val_pred)
mse_val = mean_squared_error(y_val, y_val_pred)

display(f"MAE nos dados de validação: {mae_val}")
display(f"MSE nos dados de validação: {mse_val}")
```

```python
# Criar um DataFrame para exibir os valores reais, previstos e a diferença
results_df = pd.DataFrame({
    'Actual': y_val,
    'Predicted': y_val_pred,
    'Difference': y_val - y_val_pred
})

# Formatar as colunas 'Predicted' e 'Difference' para não usar notação científica
results_df['Predicted'] = results_df['Predicted'].apply(lambda x: f'{x:.2f}')
results_df['Difference'] = results_df['Difference'].apply(lambda x: f'{x:.2f}')

display("Tabela de Resultados (Valores Reais vs. Previsões):")
display(results_df.head(20))

# Exibir a tabela de resultados
```

## Coluna ocean_proximity

```python
from sklearn.preprocessing import OneHotEncoder

df_train = pd.read_csv("/content/drive/MyDrive/2025.2/IA/Tutoriais/housePricing_train_set.csv")
df_val = pd.read_csv("/content/drive/MyDrive/2025.2/IA/Tutoriais/housePricing_val_set.csv")

# Instanciar OneHotEncoder
encoder = OneHotEncoder(handle_unknown='ignore', sparse_output=False)

# Ajustar e transformar a coluna 'ocean_proximity' nos dados de treino
ocean_proximity_train_encoded = encoder.fit_transform(df_train[['ocean_proximity']])

# Converter o array codificado para um DataFrame para facilitar a concatenação
ocean_proximity_train_df = pd.DataFrame(ocean_proximity_train_encoded, columns=encoder.get_feature_names_out(['ocean_proximity']))

display("One-Hot Encoded 'ocean_proximity' for training data:")
display(ocean_proximity_train_df.head())

# Transformar a coluna 'ocean_proximity' nos dados de validação usando o encoder ajustado
ocean_proximity_val_encoded = encoder.transform(df_val[['ocean_proximity']])

# Converter o array codificado para um DataFrame
ocean_proximity_val_df = pd.DataFrame(ocean_proximity_val_encoded, columns=encoder.get_feature_names_out(['ocean_proximity']))

display("\nOne-Hot Encoded 'ocean_proximity' for validation data:")
display(ocean_proximity_val_df.head())
```

```python
# Separar features numéricas (excluindo 'ocean_proximity' e o target)
X_train_numerical = df_train.drop(['ocean_proximity', 'median_house_value'], axis=1)
X_val_numerical = df_val.drop(['ocean_proximity', 'median_house_value'], axis=1)

# Aplicar o mesmo MinMaxScaler ajustado nos dados numéricos originais de treino
X_train_numerical_scaled = scaler.transform(X_train_numerical)
X_val_numerical_scaled = scaler.transform(X_val_numerical)

# Converter arrays numéricos escalonados para DataFrames
X_train_numerical_scaled_df = pd.DataFrame(X_train_numerical_scaled, columns=X_train_numerical.columns)
X_val_numerical_scaled_df = pd.DataFrame(X_val_numerical_scaled, columns=X_val_numerical.columns)

# Combinar features numéricas escalonadas com features categóricas one-hot encoded
X_train_processed = pd.concat([X_train_numerical_scaled_df, ocean_proximity_train_df], axis=1)
X_val_processed = pd.concat([X_val_numerical_scaled_df, ocean_proximity_val_df], axis=1)


# Separar variáveis target
y_train = df_train['median_house_value']
y_val = df_val['median_house_value']

display("Processed training features (X_train_processed):")
display(X_train_processed.head())
display("\nProcessed validation features (X_val_processed):")
display(X_val_processed.head())
display("\nTraining target (y_train):")
display(y_train.head())
display("\nValidation target (y_val):")
display(y_val.head())
```

```python
from sklearn.preprocessing import PolynomialFeatures

# Instanciar PolynomialFeatures com grau 2
poly_features = PolynomialFeatures(degree=2, include_bias=False)

# Ajustar e transformar as features de treino processadas
X_train_poly_categorical = poly_features.fit_transform(X_train_processed)

# Exibir o formato das novas features polinomiais
display("Shape of training features with polynomial and categorical features:")
display(X_train_poly_categorical.shape)
```

```python
from sklearn.linear_model import LinearRegression

# Instanciar um objeto do modelo LinearRegression
linear_poly_categorical_model = LinearRegression()

# Ajustar o modelo LinearRegression aos dados de treino transformados polinomialmente e categoricamente
linear_poly_categorical_model.fit(X_train_poly_categorical, y_train)

# Exibir uma mensagem indicando que o treino do modelo foi concluído
display("Linear Regression model with polynomial and categorical features trained successfully!")
```

```python
# Aplicar a mesma transformação PolynomialFeatures ajustada nos dados de treino às features de validação processadas
X_val_poly_categorical = poly_features.transform(X_val_processed)

# Exibir o formato das features polinomiais para dados de validação
display("Shape of validation features with polynomial and categorical features:")
display(X_val_poly_categorical.shape)
```

```python
from sklearn.metrics import mean_absolute_error, mean_squared_error

# Fazer previsões nos dados de treino com features polinomiais e categóricas
y_train_pred_poly_categorical = linear_poly_categorical_model.predict(X_train_poly_categorical)

# Calcular MAE nos dados de treino
mae_train_poly_categorical = mean_absolute_error(y_train, y_train_pred_poly_categorical)

# Calcular MSE nos dados de treino
mse_train_poly_categorical = mean_squared_error(y_train, y_train_pred_poly_categorical)

display(f"Linear Regression with Polynomial and Categorical Features MAE on training data: {mae_train_poly_categorical}")
display(f"Linear Regression with Polynomial and Categorical Features MSE on training data: {mse_train_poly_categorical}")

# Calcular MAE e MSE nos dados de treino para o modelo com features polinomiais e categóricas
```

```python
# Fazer previsões nos dados de validação com features polinomiais e categóricas
y_val_pred_poly_categorical = linear_poly_categorical_model.predict(X_val_poly_categorical)

# Calcular MAE nos dados de validação
mae_val_poly_categorical = mean_absolute_error(y_val, y_val_pred_poly_categorical)

# Calcular MSE nos dados de validação
mse_val_poly_categorical = mean_squared_error(y_val, y_val_pred_poly_categorical)

# Exibir MAE e MSE para os dados de validação
display(f"Linear Regression with Polynomial and Categorical Features MAE on validation data: {mae_val_poly_categorical}")
display(f"Linear Regression with Polynomial and Categorical Features MSE on validation data: {mse_val_poly_categorical}")

# Calcular MAE e MSE nos dados de validação para o modelo com features polinomiais e categóricas
```

```python
# Criar um DataFrame para exibir os valores reais, previstos e a diferença para o modelo polinomial com features categóricas
results_df_poly_categorical = pd.DataFrame({
    'Actual': y_val,
    'Predicted_Poly_Categorical': y_val_pred_poly_categorical,
    'Difference_Poly_Categorical': y_val - y_val_pred_poly_categorical
})

# Formatar as colunas 'Predicted_Poly_Categorical' e 'Difference_Poly_Categorical' para não usar notação científica
results_df_poly_categorical['Predicted_Poly_Categorical'] = results_df_poly_categorical['Predicted_Poly_Categorical'].apply(lambda x: f'{x:.2f}')
results_df_poly_categorical['Difference_Poly_Categorical'] = results_df_poly_categorical['Difference_Poly_Categorical'].apply(lambda x: f'{x:.2f}')

display("Tabela de Resultados do Modelo Linear com Features Polinomiais e Categóricas (Valores Reais vs. Previsões):")
display(results_df_poly_categorical.head(20))

# Exibir a tabela de resultados para o modelo com features polinomiais e categóricas
```
